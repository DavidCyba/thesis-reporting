```python
import numpy as np
import matplotlib.pyplot as plt
import matplotlib
from dataclasses import dataclass
from aquarel import load_theme
theme = load_theme("umbra_dark")
theme.apply()
```

```python
@dataclass
class Style:

    fmt="o"
    fillstyle="none"
    capsize=4
    markersize=6
    linewidth=1.5
```

DW vs DWC 1-Hop-Paths

```python


# Load data
data_dwc = np.loadtxt("./1layer/DWC-1LT2-357189.csv", delimiter=",", skiprows=3)
data_dw = np.loadtxt("./1layer/dw_data.txt")

fig, ax = plt.subplots(figsize=(8, 5), dpi=400)

ax.errorbar(
    data_dwc[:, 0],
    data_dwc[:, 1],
    yerr=data_dwc[:, 2],
    fmt=Style.fmt,
    fillstyle=Style.fillstyle,
    capsize=Style.capsize,
    markersize=Style.markersize,
    linewidth=Style.linewidth,
    label="Dirac Wilson-clover (csw = 1.0)"
)

ax.errorbar(
    data_dw[:, 0],
    data_dw[:, 1],
    yerr=data_dw[:, 2],
    fmt=Style.fmt,
    fillstyle=Style.fillstyle,
    capsize=Style.capsize,
    markersize=Style.markersize,
    linewidth=Style.linewidth,
    label="Dirac Wilson"
)

ax.set_xlabel("Mass $m$")
ax.set_ylabel("Approximation error")
ax.set_title("Approximation error  vs. mass (1 layer)")
ax.set_yscale('log')
ax.set_ylim(bottom=1e-2, top=1e0)

ax.grid(True, which="both", axis="both", alpha=0.3, ls="--")
ax.legend()

plt.tight_layout()
plt.show()
theme.apply_transforms()
```

DWC 1 layer 1-Hop-Paths vs Plaquette Paths

```python
# Load data
data_dwcp2 = np.loadtxt("./DWC_pathtype_2_vs_3/DWC-1LT2-357189.csv", delimiter=",", skiprows=3)
data_dwcp3 = np.loadtxt("./DWC_pathtype_2_vs_3/DWC-1LT3-357202.csv", delimiter=",", skiprows=3)
data_dw = np.loadtxt("./1layer/dw_data.txt")

fig, ax = plt.subplots(figsize=(8, 5), dpi=400)

ax.errorbar(
    data_dwcp2[:, 0],
    data_dwcp2[:, 1],
    yerr=data_dwcp2[:, 2],
    fmt=Style.fmt,
    fillstyle=Style.fillstyle,
    capsize=Style.capsize,
    markersize=Style.markersize,
    linewidth=Style.linewidth,
    label="DWC One-Hop-Paths"
)

ax.errorbar(
    data_dwcp3[:, 0],
    data_dwcp3[:, 1],
    yerr=data_dwcp3[:, 2],
    fmt=Style.fmt,
    fillstyle=Style.fillstyle,
    capsize=Style.capsize,
    markersize=Style.markersize,
    linewidth=Style.linewidth,
    label="DWC Circular-Paths"
)

ax.errorbar(
    data_dw[:, 0],
    data_dw[:, 1],
    yerr=data_dw[:, 2],
    fmt="^",
    fillstyle=Style.fillstyle,
    capsize=Style.capsize,
    markersize=Style.markersize,
    linewidth=Style.linewidth,
    label="Dirac Wilson"
)

ax.set_xlabel("Mass $m$")
ax.set_ylabel("Approximation error")
ax.set_title("DWC Operator Path Comparison (1 Layer, csw = 1.0)")
ax.set_yscale('log')
ax.set_ylim(bottom=1e-2, top=1e0)

ax.grid(True, which="both", axis="both", alpha=0.3, ls="--")
ax.legend()

plt.tight_layout()
plt.show()
theme.apply_transforms()
```
